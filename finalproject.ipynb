```python
from unsloth import FastLanguageModel
import torch
max_seq_length = 2048
dtype = None
load_in_4bit = True
```

```python
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/gemma-2-2b",
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)
```

```python
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj",],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",

    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
    use_rslora = False,
    loftq_config = None,
)
```

```python
alpaca_prompt = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

### Instruction:
{}

### Input:
{}

Response:
{}"""

EOS_TOKEN = tokenizer.eos_token
def formatting_prompts_func(examples):
    instructions = examples.get("instruction", [])
    inputs       = examples.get("input", [])
    outputs      = examples.get("output", [])
    texts = []
    for instruction, input, output in zip(instructions, inputs, outputs):
        text = alpaca_prompt.format(instruction, input, output) + EOS_TOKEN
        texts.append(text)
    return { "text" : texts, }
pass
```

```python
from datasets import load_dataset
dataset = load_dataset("ShenLab/MentalChat16K")
dataset = dataset.map(formatting_prompts_func, batched = True,)
dataset
```

```python
from trl import SFTTrainer
from transformers import TrainingArguments
from unsloth import is_bfloat16_supported

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset['train'],
    dataset_text_field = "text",
    max_seq_length = max_seq_length,
    dataset_num_proc = 2,
    packing = False,
    args = TrainingArguments(
        per_device_train_batch_size = 2,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,

        max_steps = 30,
        learning_rate = 2e-4,
        fp16 = not is_bfloat16_supported(),
        bf16 = is_bfloat16_supported(),
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "outputs",
        report_to = "none",
    ),
)
```

```python
trainer_stats = trainer.train()
```

```python
gpu_stats = torch.cuda.get_device_properties(0)
start_gpu_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)
max_memory = round(gpu_stats.total_memory / 1024 / 1024 / 1024, 3)
#@title Show final memory and time stats
used_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)
used_memory_for_lora = round(used_memory - start_gpu_memory, 3)
used_percentage = round(used_memory         /max_memory*100, 3)
lora_percentage = round(used_memory_for_lora/max_memory*100, 3)
print(f"{trainer_stats.metrics['train_runtime']} seconds used for training.")
print(f"{round(trainer_stats.metrics['train_runtime']/60, 2)} minutes used for training.")
print(f"Peak reserved memory = {used_memory} GB.")
print(f"Peak reserved memory for training = {used_memory_for_lora} GB.")
print(f"Peak reserved memory % of max memory = {used_percentage} %.")
print(f"Peak reserved memory for training % of max memory = {lora_percentage} %.")
```

# **Inference**

```python
def generate_answer_alpaca(query, top_k=4):
    query_embedding = embedding_model.encode(query).tolist()
    results = collection.query(query_embeddings=[query_embedding], n_results=top_k)
    retrieved_docs = "\n\n".join([
        meta.get("text", "") for meta in results["metadatas"][0] if meta.get("text")
    ])


    instruction = "Act as an empathetic mental health expert, validate the person's feelings, provide therapy and give helpful advice using the given context."
    input_text = f"Context:\n{retrieved_docs}\n\nQuestion:\n{query}"
    response_placeholder = ""
    prompt = alpaca_prompt.format(instruction, input_text, response_placeholder) + EOS_TOKEN

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=300, do_sample=True)

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

# **Gradio Implementation - Finalized**

```python
import gradio as gr

def rag_chatbot_panels(query):
    query_embedding = embedding_model.encode(query).tolist()
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=4,
        include=["metadatas", "documents", "distances"]
    )

    context_chunks = [
        meta.get("text", "") for meta in results["metadatas"][0] if meta.get("text")
    ]
    context = "\n\n".join(context_chunks)

    instruction = "You are an empathetic mental health expert currently engaged in a therapy session with your patient. Validate the patient's feelings and give helpful advice using the given context backed by sceintific data."
    input_text = f"Context:\n{context}\n\nQuestion:\n{query}"
    prompt = alpaca_prompt.format(instruction, input_text, "") + EOS_TOKEN

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=450, do_sample=True)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    similarity_scores = results.get("distances", [["N/A"]])[0]
    formatted_scores = "\n".join([f"{i+1}. {round(score, 4)}" for i, score in enumerate(similarity_scores)])

    return query, response, query, context, formatted_scores

with gr.Blocks() as interface:
    gr.Markdown("## 🧠 Mental Health Chatbot (RAG + LLM)")

    with gr.Row():
        input_box = gr.Textbox(lines=4, label="User Input")

    with gr.Row():
        response_box = gr.Textbox(label="LLM Response")
        input_echo = gr.Textbox(label="Echoed Input")

    with gr.Row():
        context_box = gr.Textbox(label="Retrieved Context")
        score_box = gr.Textbox(label="Similarity Scores")

    submit_btn = gr.Button("Generate Response")

    submit_btn.click(
        fn=rag_chatbot_panels,
        inputs=input_box,
        outputs=[input_echo, response_box, input_echo, context_box, score_box]
    )

interface.launch()
```

```python
def generate_answer_alpaca(query, context="", top_k=4):
    # If no external context is passed, use RAG from ChromaDB
    if not context:
        query_embedding = embedding_model.encode(query).tolist()
        results = collection.query(query_embeddings=[query_embedding], n_results=top_k)
        context = "\n\n".join([
            meta.get("text", "") for meta in results["metadatas"][0] if meta.get("text")
        ])

    instruction = "Act as an empathetic mental health expert. Validate the person's feelings, provide therapy, and give helpful advice using the given context."
    input_text = f"Context:\n{context}\n\nQuestion:\n{query}"
    prompt = alpaca_prompt.format(instruction, input_text, response_placeholder) + EOS_TOKEN

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_new_tokens=300, do_sample=True)

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
from datetime import datetime
```

```python
def retrieve_memory_context(query, session_id=None, top_k=3):
    query_embedding = embedding_model.encode(query).tolist()

    where_filter = {"session_id": session_id} if session_id else {}

    results = memory_collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k,
        where=where_filter,
        include=["metadatas", "documents"]
    )

    memory_contexts = []
    for meta in results["metadatas"][0]:
        if meta and "response" in meta:
            memory_contexts.append(meta["response"])

    return "\n".join(memory_contexts)
```

```python
def chat_with_memory(query, session_id, top_k=3):

    past_context = retrieve_memory_context(query, session_id=session_id, top_k=top_k)


    response = generate_answer_alpaca(query, context=past_context)
    metadata = {
        "response": response,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "session_id": session_id
    }
    memory_collection.add(
        documents=[query],
        metadatas=[metadata],
        ids=[str(uuid.uuid4())],
        embeddings=[embedding_model.encode(query).tolist()]
    )

    return response
```

```python
import gradio as gr
import uuid

def generate_session_id():
    return str(uuid.uuid4())


def gradio_chatbot(message, history, session_id):
    response = chat_with_memory(message, session_id=session_id)
    history.append((message, response))
    return history, session_id


with gr.Blocks() as demo:
    gr.Markdown("# 🧠 Empathetic AI Mental Health Assistant")
    chatbot = gr.Chatbot(label="Therapist Bot")
    msg = gr.Textbox(placeholder="Share how you're feeling...", show_label=False)
    session_state = gr.State(generate_session_id())

    msg.submit(fn=gradio_chatbot, inputs=[msg, chatbot, session_state], outputs=[chatbot, session_state])
```
